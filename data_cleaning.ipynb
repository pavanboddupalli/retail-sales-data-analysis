```python
# lets clean nnumerical data
```

```python
df['Age'].unique()
```

```python
# lets replace error values and unknown values into null values and remove the negative age

df['Age']=pd.to_numeric(df['Age'],errors='coerce')
```

```python
# df.loc[df['Age'] < 0] -- to look the negative age rows
```

```python
df = df.copy()
df = df[(df['Age'] > 0) | (df['Age'].isna())]
```

```python
df['Age'].unique()
```

```python
df['Quantity'].unique()
```

```python

# df['Quantity'] = pd.to_numeric(df['Quantity'],errors='coerce').astype('Int64')

df = df[(df['Quantity']>0) |  (df['Quantity'].isna())] 
```

```python
df['Quantity'].unique()
```

```python
df['Satisfaction'].unique()
```

```python
df['Satisfaction']=pd.to_numeric(df['Satisfaction'],errors='coerce')
```

```python
df['Satisfaction'].unique()
```

```python
# df.loc[(df['Satisfaction']<0) | (df['Satisfaction']>5)]
```

```python
df.loc[(df['Satisfaction']<1) | (df['Satisfaction']>5), 'Satisfaction']=np.nan
```

```python
df.isna().sum()
```

```python
# filling null values

df['Order_Date']=df['Order_Date'].fillna(df['Order_Date'].mode()[0])
```

```python
df['Region']=df['Region'].fillna(df['Region'].mode()[0])
```

```python
df['Sales_Channel']=df['Sales_Channel'].fillna(df['Sales_Channel'].mode()[0])
```

```python
df['Customer_Segment']=df['Customer_Segment'].fillna(df['Customer_Segment'].mode()[0])
```

```python
df['Age']=df['Age'].fillna(df['Age'].median())
```

```python
df['Category']=df['Category'].fillna(df['Category'].mode()[0])
```

```python
df['Subcategory']=df['Subcategory'].fillna(df['Subcategory'].mode()[0])
```

```python
df['Age']=df['Age'].fillna(df['Age'].median())
```

```python
# df.loc[df['Quantity'].isna()]
```

```python
df['Quantity'] = df['Quantity'].fillna( df['Total_Sales'] / df['Unit_Price']).round().astype('Int64')

# df['Quantity'] = pd.to_numeric(df['Quantity'],errors='coerce').astype('Int64')
```

```python
#  df.loc[df['Unit_Price'].isna()] 
#     to check the null values in unit price
```

```python
df['Unit_Price'] = df['Unit_Price'].fillna( df['Total_Sales'] / df['Quantity'])
```

```python
df['Payment_Method']=df['Payment_Method'].fillna(df['Payment_Method'].mode()[0])
```

```python
df['Satisfaction']=df['Satisfaction'].fillna(df['Satisfaction'].median())
```

```python
df['Delivery_Date']=df['Delivery_Date'].fillna(df['Delivery_Date'].mode()[0])
```

```python
df=df.dropna()
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
df.to_csv("cleaned_data.csv",index=False)
```
